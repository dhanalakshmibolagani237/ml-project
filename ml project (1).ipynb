```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix,ConfusionMatrixDisplay
data=pd.read_csv("D:\disease_prediction_dataset.csv")
print(data)
```

```python
x=data.iloc[:,:-1]
print(x)
```

```python
y=data.iloc[:,-1]
print(y)
```

```python
x=data.drop("Disease",axis=1)
y=data["Disease"]
print(x)
print(y)
```

```python
x_train,x_test,y_train,y_test=train_test_split(
    x,y,test_size=0.2,random_state=42
)
```

```python
model=DecisionTreeClassifier(random_state=42)
model.fit(x_train,y_train)
```

```python
pred=model.predict(x_test)
print(pred)
```

```python
print("Accuracy:",model.score(x_test,y_test))
```

```python
model=LogisticRegression(max_iter=1000)
model.fit(x_train,y_train)
```

```python
pred=model.predict(x_test)
print(pred)
print("Accuracy:",model.score(x_test,y_test))
```

```python
model=KNeighborsClassifier(n_neighbors=5)
model.fit(x_train,y_train)
```

```python
pred=model.predict(x_test)
print(pred)
```

```python
print("Accuracy:",model.score(x_test,y_test))
```

```python
new_patient=[[1,1,1,1,1,1,1]]
print(model.predict(new_patient))
```

```python
dt=DecisionTreeClassifier(random_state=42)
rf=RandomForestClassifier(n_estimators=100,random_state=42)
lr=LogisticRegression(max_iter=1000)
knn=KNeighborsClassifier(n_neighbors=5)
dt.fit(x_train,y_train)
rf.fit(x_train,y_train)
lr.fit(x_train,y_train)
knn.fit(x_train,y_train)
dt_accuracy=dt.score(x_test,y_test)
rf_accuracy=rf.score(x_test,y_test)
lr_accuracy=lr.score(x_test,y_test)
knn_accuracy=knn.score(x_test,y_test)
models=["Decision Tree","randomForest","LogisticRegression","KNN"]
accuracy=[dt_accuracy,rf_accuracy,lr_accuracy,knn_accuracy]
plt.bar(models,accuracy)
plt.xlabel("models")
plt.ylabel("Accuracy")
plt.title("Decision Prediction")
plt.xticks(rotation=20)
plt.show()
```

```python
dt_pred=dt.predict(x_test)
rf_pred=rf.predict(x_test)
lr_pred=lr.predict(x_test)
knn_pred=knn.predict(x_test)
cm_dt=confusion_matrix(y_test,dt_pred)
cm_rf=confusion_matrix(y_test,rf_pred)
cm_lr=confusion_matrix(y_test,lr_pred)
cm_knn=confusion_matrix(y_test,knn_pred)
print("Decision  Tree confusion matrix:")
print(cm_dt)
print("Random Forest Confusion Matrix:")
print(cm_rf)
print("Logistic Regression Confusion Matrix:")
print(cm_lr)
print("Knn Confusion Matrix:")
print(cm_knn)
```

```python
ConfusionMatrixDisplay.from_predictions(y_test,dt_pred)
plt.title("Decision Tree confusion matrix")
plt.show()
ConfusionMatrixDisplay.from_predictions(y_test,rf_pred)
plt.title("Random Forest confusion matrix")
plt.show()
ConfusionMatrixDisplay.from_predictions(y_test,lr_pred)
plt.title("Logistic Rregression confusion matrix")
plt.show()
ConfusionMatrixDisplay.from_predictions(y_test,knn_pred)
plt.title("KNN confusion matrix")
plt.show()
```
